```python
#Accessing:
numbers = [10,20,30,40]
print(numbers[0])
print(numbers[1])
```

```python
#slicing:
print(numbers[0:2])
print(numbers[1:3])
print(numbers[-2])
print(numbers[::-1])
print(numbers[0:5:2])
```

```python
# list concatination:
lst1 =[1,2,3,4]
lst2 =[3,4,5,6,7]
lst3 =lst1+lst2
print(lst3)
```

```python
#list repation:
fruits =["apple","banana","cherry"]
repation_list = fruits * 2
print(repation_list)
```

```python
#list methods
## append()
l = [1, 2,3,5.5,'a','b', 1,2,3]
print(l)
l.append(100)
print("After append Operation:", l)
```

```python
## extend()
l = [1, 2,3,4,5]
l.extend([200,300])
print(l[5])
print("After extend Operation:", l)
```

```python
## insert()
l = [1, 2,3,5.5,]
l.insert(40,10)
print("After insert Operation:", l)
```

```python
# remove(), pop(3), clear(), del
l = [1, 2, 3, 7.8, 'meghana', 'a', 'b', 1, 2, 3]
print("Original list:", l)
l.remove(3)
print("After remove operation:", l)
l.pop()
print("After pop operation:", l)
l.pop(3)
print("After pop with position:", l)
del l[0]
print(l)
del l[2:5]
print("After del operation:",l)
l.clear()
print("After clear opration:",l)
```
